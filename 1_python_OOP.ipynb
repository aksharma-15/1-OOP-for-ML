## That's where the usage of private variable comes.

```python
class BankAccount:
  account_number = 101 # this is to make sure that account numbers always stay unique

  def __init__(self, name, amount):
    self.name = name
    self.acc_no = BankAccount.account_number

    BankAccount.account_number += 1
    self.__amount = amount # --> pvt variable
    print(f"{self.name}'s new account is opened")

  def __repr__(self):
    return f"{self.name}, with account num: {self.acc_no} has Rs.{self.__amount}" # __repr__ is a dunder method used tp display desired output when we type print(object_name)

  def get_balance(self):
    return f"Your balace Rs.{self.__amount}"

  def withdrawl(self, amt):
    if amt > self.__amount:
      return 'Insufficient funds'

    self.__amount -= amt
    print(f"Withdrawing cash Rs.{amt}")
    print(f"Balance remaining Rs.{self.__amount}")

  def deposit(self, amt):
    if amt <= 0:
      return 'Enter a value to deposit'

    self.__amount += amt
    print(f"Deposited cash Rs.{amt}")
    print(f"Updated balance Rs.{self.__amount}")
```

```python
aks_acc = BankAccount('Abhay', 10000)
goldie_acc = BankAccount('Himanshu', 15000)
```

## Account merging won't work unless we hardcode an 'add' or 'merge' function inside the class.

```python
class BankAccount:
  account_number = 101 # this is to make sure that account numbers always stay unique

  def __init__(self, name, amount):
    self.name = name
    self.acc_no = BankAccount.account_number

    BankAccount.account_number += 1
    self.__amount = amount # --> pvt variable
    print(f"{self.name}'s new account is opened")

  def __repr__(self):
    return f"{self.name}, with account num: {self.acc_no} has Rs.{self.__amount}" # __repr__ is a dunder method used tp display desired output when we type print(object_name)

  def get_balance(self):
    return f"Your balace Rs.{self.__amount}"

  def withdrawl(self, amt):
    if amt > self.__amount:
      return 'Insufficient funds'

    self.__amount -= amt
    print(f"Withdrawing cash Rs.{amt}")
    print(f"Balance remaining Rs.{self.__amount}")

  def deposit(self, amt):
    if amt <= 0:
      return 'Enter a value to deposit'

    self.__amount += amt
    print(f"Deposited cash Rs.{amt}")
    print(f"Updated balance Rs.{self.__amount}")

  def __add__(self, other):
    new_name = f'{self.name} + {other.name}'
    new_balance = f'{self.__amount} + {other.__amount}'
    return BankAccount(new_name, new_balance)
```

# Inheritance

```python
# parent class

class BankAccount:
  account_number = 101 # this is to make sure that account numbers always stay unique

  def __init__(self, name, amount):
    self.name = name
    self.acc_no = BankAccount.account_number

    BankAccount.account_number += 1
    self.__amount = amount # --> pvt variable
    print(f"{self.name}'s new account is opened")

  def __repr__(self):
    return f"{self.name}, with account num: {self.acc_no} has Rs.{self.__amount}" # __repr__ is a dunder method used tp display desired output when we type print(object_name)

  def get_balance(self):
    return f"Your balace Rs.{self.__amount}"

  def withdrawl(self, amt):
    if amt > self.__amount:
      return 'Insufficient funds'

    self.__amount -= amt
    print(f"Withdrawing cash Rs.{amt}")
    print(f"Balance remaining Rs.{self.__amount}")

  def deposit(self, amt):
    if amt <= 0:
      return 'Enter a value to deposit'

    self.__amount += amt
    print(f"Deposited cash Rs.{amt}")
    print(f"Updated balance Rs.{self.__amount}")

  def __add__(self, other):
    new_name = f'{self.name} + {other.name}'
    new_balance = f'{self.__amount} + {other.__amount}'
    return BankAccount(new_name, new_balance)
```

## Creating a child class to demonstrate the usage of inheritance. We will make a Savings account out of the already existing BankAccount class.

```python
# child class --> inherits the methods and attributes used inside the parent class
# We can also make new attributes and methods inside a child class

class SavingsAccount(BankAccount):
  def __init__(self, name, amount):
    super().__init__(name, amount) # calling the parent __init__
    self.__amount = amount
    self.__interest_rate = 0.035 # 3.5 percent interest p.a
    self.max_transac_limit = 10
    self.current_trnsc = 0

  def calculate_interest(self):
    interest = self.__amount * self.__interest_rate
    return f"Your total interest Rs. {interest:0.2f}"
```

```python
abhay_sv = SavingsAccount('Abhay', 10000)
```

```python
print(abhay_sv)
```

```python
abhay_sv.calculate_interest()
```
